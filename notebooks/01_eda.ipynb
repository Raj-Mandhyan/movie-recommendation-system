```python
def convert(text):
    L = []

    for i in ast.literal_eval(text):
        L.append(i["name"])

    return L
```

```python
convert(movies.iloc[0]["genres"])
```

```python
movies["genres"] = movies["genres"].apply(convert)
```

```python
def convert3(text):
    L = []

    counter = 0

    for i in ast.literal_eval(text):
        if counter != 3:
            L.append(i["name"])
            counter += 1
        else:
            break

    return L
```

```python
convert3(movies.iloc[0]["cast"])
```

```python
movies["cast"] = movies["cast"].apply(convert3)
```

```python
movies.head()
```

```python
ast.literal_eval(movies.iloc[0]["crew"])
```

```python
def fetch_director(text):
    L = []

    for i in ast.literal_eval(text):
        if i["job"] == "Director":
            L.append(i["name"])
            break

    return L
```

```python
fetch_director(movies.iloc[0]["crew"])
```

```python
movies["crew"] = movies["crew"].apply(fetch_director)
```

```python
movies.head()
```

```python
def collapse(L):
    return [i.replace(" ", "") for i in L]
```

```python
collapse(["Sam Worthington", "Science Fiction", "James Cameron"])
```

```python
movies["cast"] = movies["cast"].apply(collapse)

movies["crew"] = movies["crew"].apply(collapse)

movies["genres"] = movies["genres"].apply(collapse)

movies["keywords"] = movies["keywords"].apply(collapse)
```

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()
```

```python
def stem(text):
    y = []

    for i in text.split():
        y.append(ps.stem(i))

    return " ".join(y)
```

```python
stem(new_df["tags"].iloc[0])
```

```python
new_df["tags"] = new_df["tags"].apply(stem)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer(
    max_features=5000,
    stop_words="english"
)
```

```python
vectors = cv.fit_transform(new_df["tags"]).toarray()
```

```python
vectors.shape
```

```python
cv.get_feature_names_out()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
```

```python
def recommend(movie):
    # Find the index of the movie
    movie_index = new_df[new_df["title"] == movie].index[0]

    # Get similarity scores
    distances = similarity[movie_index]

    # Pair movie index with similarity score
    movies_list = sorted(
        list(enumerate(distances)),
        reverse=True,
        key=lambda x: x[1]
    )

    # Print top 5 recommendations
    for i in movies_list[1:6]:
        print(new_df.iloc[i[0]].title)
```
